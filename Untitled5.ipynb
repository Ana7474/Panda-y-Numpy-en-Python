# Reto - Análisis del progreso mundial de vacunación

## Introducción

La pandemia de COVID-19 cambió por completo nuestra rutina y puso a prueba a los sistemas de salud de todo el mundo. Sin las vacunas, hubiera sido imposible acercarnos de nuevo a eso que conocemos como “normalidad”.

En este proyecto trabajaré con el archivo "country_vaccinations.csv", que contiene registros diarios de vacunación por país. Usando Python y pandas analizaré estos datos para entender cuántas vacunas se aplicaron, cómo se distribuyeron y qué papel tuvieron las distintas marcas en el panorama global.

## Objetivo

Aplicar Python y pandas para explorar un conjunto de datos real y obtener resúmenes claros e interpretables que puedan revisarse y compartirse en un archivo de Excel.



### a) Extraer la información del archivo.

Utilicé la función "read_csv" de pandas para leer el archivo "country_vaccinations.csv" y cargarlo en un dataframe llamado "vac". Además, mostré las primeras filas para corroborar que este se haya leído correctamente.

```python
# a) Extraer la información del archivo.
import pandas as pd

vac = pd.read_csv('/content/country_vaccinations.csv')
vac.head()
```

### b) Mostrar la estructura y tipos de datos de cada columna para identificar qué operaciones puedes realizar con cada una de ellas, asegurándote que las columnas con fechas sean del tipo datetime64.

Convertí la columna "date" a tipo de fecha con "pd.to_datetime" y usé "info()" para revisar la estructura del dataframe y confirmar que la conversión fuera correcta.*italicized text*

```python
# b) Mostrar la estructura y tipos de datos de cada columna para identificar qué operaciones puedes realizar con cada una de ellas, asegurándote que las columnas con fechas sean del tipo datetime64.

# Convertir la columna "date" a tipo fecha
vac['date'] = pd.to_datetime(vac['date'])

# Revisar la estructura y tipos de datos
print('Estructura del dataframe vac:')
vac.info()
```

### c) Determinar la cantidad de vacunas aplicadas de cada compañía

En este inciso conté cuántas vacunas se aplicaron en total por cada tipo en la columna "vaccines".  
Para hacerlo, agrupé el dataframe por "vaccines" con "groupby" y sumé la columna "daily_vaccinations". Obteniendo así un resumen del total de aplicaciones diarias de cada tipo.

```python
# c) Determinar la cantidad de vacunas aplicadas de cada compañía

# Agrupar por "vaccines" y sumar
vacunas_por_compania = vac.groupby('vaccines')['daily_vaccinations'].sum()

# Mostrar el resumen
vacunas_por_compania
```

### d) Obtener la cantidad de vacunas aplicadas en todo el mundo.

Para mostrar cuantas vacunas se aplicaron en total en todo el mundo sumé la columna "daily_vaccinations" con "sum()" y obtuve el total de aplicaciones registradas en el archivo.


```python
# d) Obtener la cantidad de vacunas aplicadas en todo el mundo.

total_mundial = vac['daily_vaccinations'].sum()
print('Vacunas aplicadas en todo el mundo:', total_mundial)
```

### e) Calcular el promedio de vacunas aplicadas por país.

Para calcular el promedio diario de vacunas aplicadas en cada país agrupé los datos por "country" y saqué la media de "daily_vaccinations" con "mean()".

```python
# e) Calcular el promedio de vacunas aplicadas por país.

promedio_por_pais = vac.groupby('country')['daily_vaccinations'].mean()
promedio_por_pais
```

### f) Determinar la cantidad de vacunas aplicadas el día 29/01/21 en todo el mundo.

Filtré en la columa "date" por "2021-01-29" y, sobre ese subconjunto, sumé la columna "daily_vaccinations" para obtener el total mundial de ese día.


```python
# f) Determinar la cantidad de vacunas aplicadas el día 29/01/21 en todo el mundo.

vac_29ene = vac[vac['date'] == '2021-01-29']
total_29ene = vac_29ene['daily_vaccinations'].sum()

print('Vacunas aplicadas el día 29/01/2021 en todo el mundo:', total_29ene)
```

### g) Crear un dataframe nuevo denominado conDiferencias que contenga los datos originales y una columna derivada (diferencias) con las diferencias de aplicación entre las columnas daily_vaccionations y daily_vaccionations_raw.  

Creé un nuevo dataframe llamado "conDiferencias" a partir de "vac". Le agregué una columna llamada "diferencias", la cual es una resta entre "daily_vaccinations" y "daily_vaccinations_raw", esto con el fin de comparar el valor ajustado de vacunaciones diarias contra el valor crudo reportado.

```python
# g) Crear un dataframe nuevo denominado conDiferencias que contenga los datos originales y una columna derivada (diferencias) con las diferencias de aplicación entre las columnas daily_vaccionations y daily_vaccionations_raw.
conDiferencias = vac
conDiferencias['diferencias'] = (
    conDiferencias['daily_vaccinations'] - conDiferencias['daily_vaccinations_raw']
)

conDiferencias.head()
```

### h) Obtener el periodo de tiempo entre el registro con fecha más reciente y el registro con fecha más antigua.

Obtuve la fecha más antigua con "min()" sobre la columna "date", la más reciente con "max()", luego resté ambas fechas para calcular cuántos días abarca el conjunto de datos.


```python
# h) Obtener el periodo de tiempo entre el registro con fecha más reciente y el registro con fecha más antigua.

fecha_min = vac['date'].min()
fecha_max = vac['date'].max()
periodo = fecha_max - fecha_min

print('Fecha más antigua:', fecha_min)
print('Fecha más reciente:', fecha_max)
print('Período total:', periodo)
```

### i) Crear un dataframe nuevo denominado conCantidad que contenga los datos originales y una columna derivada (canVac) con la cantidad de vacunas utilizadas cada día (usar la columna vaccines y separar por el carácter , ).

Creé un dataframe llamado "conCantidad" a partir de "vac" y le agregué una columna llamada "canVac".  
Obtuve la columna a partir de "vaccines", dividiendo el texto por comas y contando cuántos elementos quedan. Esto nos dice cuantas vacunas distintas hay.


```python
# i) Crear un dataframe nuevo denominado conCantidad que contenga los datos originales y una columna derivada (canVac) con la cantidad de vacunas utilizadas cada día (usar la columna vaccines y separar por el carácter , ).

conCantidad = vac
conCantidad['canVac'] = conCantidad['vaccines'].str.split(',').str.len()

# Revisar algunas filas para comprobar el resultado
conCantidad[['country', 'date', 'vaccines', 'canVac']].head()
```

### j) Generar un dataframe denominado antes20 con todos los registros que se hayan realizado antes del 20 de diciembre de 2020.

Filtré los datos para quedarme solo con los registros anteriores al 20 de diciembre de 2020.
A partir del dataframe "conCantidad", seleccioné las filas donde "date < '2020-12-20'" y guardé ese subconjunto en un nuevo dataframe llamado "antes20".


```python
# j) Generar un dataframe denominado antes20 con todos los registros que se hayan realizado antes del 20 de diciembre de 2020.

antes20 = conCantidad[conCantidad['date'] < '2020-12-20']

# Revisar algunas filas para comprobar el resultado
antes20.head()
```

### k) Obtener un dataframe denominado pfizer con todos los registros donde se haya utilizado la vacuna Pfizer.

Separé todos los registros en los que se utilizó la vacuna Pfizer filtrando el dataframe "conCantidad" y buscando las filas donde la columna "vaccines" contenía la palabra "Pfizer". Guardé ese subconjunto en un nuevo dataframe llamado "pfizer".

```python
# k) Obtener un dataframe denominado pfizer con todos los registros donde se haya utilizado la vacuna Pfizer.

pfizer = conCantidad[conCantidad['vaccines'].str.contains('Pfizer', na=False)]

# Revisar algunas filas para comprobar el resultado
pfizer.head()
```

### l) Almacenar los dataframes generados (conDiferencias, conCantidad, antes20 y pfizer) en un archivo de Excel denominado resultadosReto.xlsx, donde cada dataframe ocupe una hoja diferente. Se recomienda ver la documentación de pd.ExcelWriter.

Reuní todos los resultados en un solo archivo utilizando "pd.ExcelWriter" para crear el archivo "resultadosReto.xlsx". Este archivo contiene:

Los dataframes "conDiferencias", "conCantidad", "antes20" y "pfizer", en sus respectivas hojas.

Así, toda la información  queda en un solo archivo de Excel que se puede revisar y utilizar fácilmente.


```python
# l) Almacenar los dataframes generados en un archivo de Excel denominado resultadosReto.xlsx.

ruta_salida = '/content/resultadosReto.xlsx'

with pd.ExcelWriter(ruta_salida) as writer:
    conDiferencias.to_excel(writer, sheet_name='conDiferencias', index=False)
    conCantidad.to_excel(writer, sheet_name='conCantidad', index=False)
    antes20.to_excel(writer, sheet_name='antes20', index=False)
    pfizer.to_excel(writer, sheet_name='pfizer', index=False)

ruta_salida
```
